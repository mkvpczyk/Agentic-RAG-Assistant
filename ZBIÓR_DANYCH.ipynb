```python
# Import bibliotek
import pandas as pd
from datasets import load_dataset
```

```python
# Załadowanie zbioru danych
df = load_dataset("S-Labs/prompt-injection-dataset")
```

```python
# Konwersja zbioru danych
df_train = pd.DataFrame(df["train"])
```

```python
# Sprawdzenie długości zbioru
len(df_train)
```

```python
# Sprawdzenie pustych wierszy
df_train['text'].isnull().sum()
```

```python
# Zliczenie duplikatów
duplicates = df_train.duplicated(subset=['text']).sum()
```

```python
# Rozkład klas
classes = df_train['label'].value_counts(normalize=True) * 100
```

```python
# Rozkład klas
classes.to_string()
```

```python
# Wylicza liczbę słów w każdym prompcie
df_train['word_count'] = df_train['text'].apply(lambda x: len(str(x).split()))
```

```python
# Wyciągnięcie krótkich promptów
short = df_train[df_train['word_count'] <= 2]
```

```python
# Ilość krótkich promptów
len(short)
```

```python
# Rozkład klas dla krótkich promptów
if len(short) > 0:
    print(short['label'].value_counts())
```
